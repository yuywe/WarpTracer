## Batched poses

Each batch member sees the same static geometry. Sensor buffers are reused;
calling `scan()` again overwrites the previous device outputs from that sensor.
The explicit `.numpy()` call synchronizes and copies for inspection.


```python
import numpy as np
from racesense3d import demo_scene, camera, CAMERA_TO_BODY
scene = demo_scene(DEVICE)
sensor = scene.sensor(camera(160,120), batch_size=4)
positions = np.array([[0,0,1], [1,0,1], [2,-2,1], [0,2,1]], dtype=np.float32)
scan = sensor.scan(positions, CAMERA_TO_BODY)
depth, valid = scan.numpy()
assert depth.shape == (4,120,160)
fig, axes = plt.subplots(1,4,figsize=(14,3),layout="constrained")
for i, ax in enumerate(axes):
    ax.imshow(np.ma.array(depth[i],mask=~valid[i]),vmin=0,vmax=12,cmap="viridis")
    ax.set_title(f"Pose {i}: {positions[i].tolist()}")
    ax.axis("off")
plt.show()
```
